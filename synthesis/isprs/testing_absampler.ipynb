```python
import matplotlib.pyplot as plt
import matplotlib.tri as tri
import numpy as np
import torch

class TernaryPlotter:
    def __init__(self, labels=('GV', 'NPV', 'Soil')):
        """
        Handles projection and plotting on a simplex.
        Labels correspond to tensor columns [0, 1, 2].
        """
        self.labels = labels
        
        # Define vertices of the equilateral triangle
        # Order: Top (Col 0), Left (Col 1), Right (Col 2)
        self.top = np.array([0.5, np.sqrt(3)/2])
        self.left = np.array([0.0, 0.0])
        self.right = np.array([1.0, 0.0])
        
        # Stacking for easy matrix multiplication
        # Shape (3, 2) -> Rows correspond to Col 0, Col 1, Col 2
        self.vertices = np.vstack([self.top, self.left, self.right])

    def project(self, abundances):
        """
        Project (N, 3) abundances to (N, 2) Cartesian coordinates.
        Expects numpy array or torch tensor.
        """
        if isinstance(abundances, torch.Tensor):
            abundances = abundances.detach().cpu().numpy()
            
        # Matrix multiplication: (N, 3) @ (3, 2) -> (N, 2)
        # This automatically performs the weighted sum of vertex coordinates
        coords = abundances @ self.vertices
        return coords

    def draw_boundary(self, ax):
        """Draws the triangle boundary and labels."""
        # Draw the triangle lines
        triangle = np.vstack([self.top, self.left, self.right, self.top])
        ax.plot(triangle[:, 0], triangle[:, 1], 'k-', lw=1.5)
        
        # Label the vertices (Pure Endmembers)
        offset = 0.05
        ax.text(self.top[0], self.top[1] + offset, self.labels[0], ha='center', fontsize=12, fontweight='bold')
        ax.text(self.left[0] - offset, self.left[1] - offset/2, self.labels[1], ha='center', fontsize=12, fontweight='bold')
        ax.text(self.right[0] + offset, self.right[1] - offset/2, self.labels[2], ha='center', fontsize=12, fontweight='bold')
        
        # Clean up axes
        ax.axis('off')
        ax.set_aspect('equal')

    def plot(self, abundances, kind='scatter', ax=None, title=None, **kwargs):
        if ax is None:
            fig, ax = plt.subplots(figsize=(8, 8))
            
        coords = self.project(abundances)
        self.draw_boundary(ax)
        
        if kind == 'scatter':
            # Low alpha is crucial for visualizing density in scatter plots
            ax.scatter(coords[:, 0], coords[:, 1], s=1, alpha=0.5, c='darkred', **kwargs)
            
        elif kind == 'density':
            # Compute Kernel Density Estimate (KDE)
            # This reproduces the "Heatmap" style plots
            from scipy.stats import gaussian_kde
            
            # Filter NaNs just in case
            valid_mask = np.isfinite(coords).all(axis=1)
            x, y = coords[valid_mask].T
            
            # Fit KDE
            k = gaussian_kde(np.vstack([x, y]))
            
            # Create a grid over the triangle bounds
            xi, yi = np.mgrid[0:1:200j, 0:np.sqrt(3)/2:200j]
            zi = k(np.vstack([xi.flatten(), yi.flatten()]))
            
            # Mask out points outside the triangle to keep edges clean
            # Barycentric check: sum of weights > 1 or < 0
            # Inverse project grid points to barycentric
            # Simple check: is point inside hull?
            triang = tri.Triangulation([0, 1, 0.5], [0, 0, np.sqrt(3)/2])
            mask = tri.LinearTriInterpolator(triang, np.ones(3))(xi, yi)
            zi = zi.reshape(xi.shape)
            zi[mask.mask] = np.nan
            
            ax.contourf(xi, yi, zi, levels=50, cmap='viridis', **kwargs)

        if title:
            ax.set_title(title, y=1.05, fontsize=14)
            
        return ax
```
