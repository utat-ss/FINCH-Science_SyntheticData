# FINCH unmixing — aggregated results (seed 42)

This notebook merges the CNN+FNO scaling summary with the MLP summary, formats the final table, and exports:
- `all_models_summary_seed42.csv`
- `all_models_summary_seed42.html`


```python
import pandas as pd

cnn_fno_df = pd.read_csv("../results/cnn_fno_scaling_summary_seed42.csv")
mlp_df = pd.read_csv("~/Documents/Projects/UTAT/Finch-Science_SyntheticData/onboarding/Sam/mlp_summary_seed42.csv")
```

```python
final_df = pd.concat([cnn_fno_df, mlp_df], ignore_index=True)
final_df = final_df.sort_values(["model", "size"], kind="stable")
final_df["epochs_planned"] = final_df["epochs_planned"].fillna(80).astype(int)
final_df
```

```python
preferred_order = [
    "model", "size", "run_dir",
    "best_epoch", "epochs_ran", "epochs_planned",
    "train_mse_at_best", "val_mse_at_best", "best_val_mse", "test_mse",
    "val_r2", "test_r2",
    "generalization_gap",
    "seed"
]

# keep only cols that exist, then append any extras at end
ordered_cols = [c for c in preferred_order if c in final_df.columns] + [c for c in final_df.columns if c not in preferred_order]
final_view = final_df[ordered_cols]

final_view.style.format({
    "train_mse_at_best": "{:.6f}",
    "val_mse_at_best": "{:.6f}",
    "best_val_mse": "{:.6f}",
    "test_mse": "{:.6f}",
    "val_r2": "{:.4f}",
    "test_r2": "{:.4f}",
    "generalization_gap": "{:.6f}",
}).hide(axis="index")
```

```python
best2_by_size = (
    final_df[final_df["size"].isin(["L","M","XL"])]
    .sort_values(["size","test_mse"], ascending=[True, True])
    .groupby("size", as_index=False)
    .head(2)[["size","model","test_mse","test_r2","best_epoch","epochs_ran","run_dir"]]
)

best2_by_size.style.format({"test_mse":"{:.6f}", "test_r2":"{:.4f}"}).hide(axis="index")
```

```python
import pandas as pd
from pathlib import Path

SEED = 42

cnn_fno_params = pd.read_csv(Path(f"../results/fno_cnn_params_seed{SEED}.csv"))
mlp_params = pd.read_csv(Path("~/Documents/Projects/UTAT/FINCH-Science_SyntheticData/onboarding/Sam/mlp_params_seed42.csv").expanduser())

params_df = pd.concat([cnn_fno_params, mlp_params], ignore_index=True)

final_df = final_df.merge(params_df[["model","size","n_params"]], on=["model","size"], how="left")
# final_df.to_csv("model_params_seed42.csv", index=False)
```
